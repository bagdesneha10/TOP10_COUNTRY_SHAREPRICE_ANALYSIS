```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
a=pd.read_csv("../Datasets/Arju_Bank.xls")
b=pd.read_csv("../Datasets/Anupam_USA_Final.csv")
c=pd.read_csv("../Datasets/divya_HK_final.csv")
d=pd.read_csv("../Datasets/Janhavi_Indian_Automobile.xls")
e=pd.read_csv("../Datasets/Ninad_USA_Final.csv")
f=pd.read_csv("../Datasets/sneha_HK_final.csv")
```

```python
df=pd.concat([a,b,c,d,e,f],axis=0)
```

```python
df.Date = df.Date.astype('datetime64[ns]')
```
